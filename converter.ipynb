```python
import yaml
import json
import csv

with open("globs/config.yaml", "r") as f:
    data = yaml.safe_load(f)
    
count = len(data['items'])
max_globs = max(len(item['glob']) for item in data['items'])


# JSON
with open("globs/config.json", "w") as f:
    json.dump(data, f, indent=4)


# TXT lines
with open("globs/config.txt", "w") as f:
    for item in data['items']:
        for glob in item['glob']:
            f.write(glob + "\n")


# CSV
with open("globs/config.csv", "w") as f:
    w = csv.writer(f)
    w.writerow(['name', 'type'] + [f'glob {i+1}' for i in range(max_globs)])
    for item in data['items']:
        w.writerow([item['name'], item['type']] + [glob for glob in item['glob']])


# Toml
with open("globs/config.toml", "w") as f:
    f.write("[items]\n")
    for item in data['items']:
        f.write(f'[[items]]\n')
        f.write(f'name = "{item["name"]}"\n')
        f.write(f'type = "{item["type"]}"\n')
        f.write(f'glob = [\n')
        for glob in item['glob']:
            f.write(f'    "{glob}",\n')
        f.write(f']\n\n')
```
